Quick script to run over all UIUC airfoils to find the highest CL value at given Mach number and Reynolds Number.

```python
import math

import numpy as np
from pathlib import Path

from compute_L_and_D import compute_cl
```

Loop over all available airfiols.

```python
airfoil_list = []
X_list = []
airfoil_names_file_path = Path('./airfoil_utils/airfoil_names.txt')
airfoil_coordinates_dir_path = Path('./airfoil_utils/airfoils')
with open(airfoil_names_file_path, 'r') as air_f:
    for line in air_f:
        airfoil_name = line.strip()[:-4]
        airfoil_list.append(airfoil_name)
        X = np.loadtxt(airfoil_coordinates_dir_path / f'{airfoil_name}.dat')
        X_list.append(X)

# print(airfoil_list)
# print(X_list)
```

For each airfoil run XFoil for find CL.

```python
CL_list = []

Re = 1e6
M = 0.1
angle_of_attack = 0
max_iter = 100

for X in X_list:
    CL = compute_cl(X = X, M = M, Re = Re, angle_of_attack = angle_of_attack, max_iter = max_iter)
    CL_list.append(CL)
```

Find the maximum CL.

```python
print(f'Max CL in the database: {max(CL_list)}')
```

Looks like there are outliers. Also `nan`s.

```python
CL_list_clean = [CL for CL in CL_list if not math.isnan(CL)]
```
